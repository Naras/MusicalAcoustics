```python
def play_audio_timestamp(*args):
    """Loads and plays a specific window of a long audio file."""
    # print(f'no. of {len(args)} args {args}')
    path = args[0]
    start_min, start_sec, end_min, end_sec = args[1] 
    nam = args[2] if len(args) > 2 else "Unknown"
    # print(f'{start_min}:{start_sec} {end_min}:{end_sec} {nam}')
    t_start = (start_min * 60) + start_sec
    t_end = (end_min * 60) + end_sec
    t_duration = t_end - t_start
    
    # Load segment
    y_seg, sr_seg = librosa.load(path, offset=t_start, duration=t_duration)
    
    print(f"Playing snippet {nam} from {start_min}:{start_sec:02f} to {end_min}:{end_sec:02f}")
    return ipd.Audio(y_seg, rate=sr_seg)
```

# Swara Concatenation and Fidelity Analysis

In this section, we implement a concatenator for the individual swaras in the teaching audio. We will join single swara slices (e.g., `sa`, `ri1`, `ga3`, `ma1`) and compare them to the original continuous phrases to evaluate the fidelity of the concatenated sequences.

To prevent click/pop sounds at the boundaries of concatenated segments, we also implement a **crossfade** option which smoothly fades out the end of the previous note while fading in the start of the next note.

```python
import numpy as np
import matplotlib.pyplot as plt
import librosa
import librosa.display
import IPython.display as ipd
from IPython.display import display

def concatenate_swaras(y, sr, swara_list, crossfade_ms=15):
    slices = []
    for t in swara_list:
        t_start = (t[0] * 60) + t[1]
        t_end = (t[2] * 60) + t[3]
        idx_start = int(t_start * sr)
        idx_end = int(t_end * sr)
        slices.append(y[idx_start:idx_end])
        
    if not slices:
        return np.array([]), None
        
    fade_len = int((crossfade_ms * sr) / 1000)
    y_concat = slices[0].copy()
    
    for next_slice in slices[1:]:
        if fade_len > 0 and len(y_concat) >= fade_len and len(next_slice) >= fade_len:
            fade_out = np.linspace(1.0, 0.0, fade_len)
            fade_in = np.linspace(0.0, 1.0, fade_len)
            y_concat[-fade_len:] = y_concat[-fade_len:] * fade_out
            faded_next = next_slice.copy()
            faded_next[:fade_len] = faded_next[:fade_len] * fade_in
            overlap = y_concat[-fade_len:] + faded_next[:fade_len]
            y_concat = np.concatenate([y_concat[:-fade_len], overlap, faded_next[fade_len:]])
        else:
            y_concat = np.concatenate([y_concat, next_slice])
            
    return y_concat, ipd.Audio(y_concat, rate=sr)

def compare_many_versions(y, sr, original_tuple, algos, crossfade_ms=15, title="", descriptions=[]):
    # Extract original
    t_start_orig = original_tuple[0] * 60 + original_tuple[1]
    t_end_orig = original_tuple[2] * 60 + original_tuple[3]
    y_orig = y[int(t_start_orig * sr) : int(t_end_orig * sr)]
    
    
    print(f"[{title}] Original Duration: {len(y_orig)/sr:.3f} seconds")
    print(f"1. Original Phrase:")
    display(ipd.Audio(y_orig, rate=sr))
    y_algos = ()
    i = -1
    for swara_list_algo in algos:
        i += 1
        description = descriptions[i] if i <= len(descriptions) else "I don't know"
        y_algo, _ = concatenate_swaras(y, sr, swara_list_algo, crossfade_ms=crossfade_ms)
        print(f"[{title}] {description} Duration: {len(y_algo)/sr:.3f} seconds")    
        display(ipd.Audio(y_algo, rate=sr))
        y_algos+=(y_algo,)
        
    # print(f"[{title}] Algo 3 (Manual Guided - No Constraints) Duration: {len(y_algo3)/sr:.3f} seconds")
    # print(f"[{title}] Algo 1 (Manual Guided - With Constraints) Duration: {len(y_algo1)/sr:.3f} seconds")
    
        # print(f"[{title}] Algo 2 (Fully Auto - Medians) Duration: {len(y_algo)/sr:.3f} seconds")
    
    # print(f"3. Algo 3 (Manual Guided - No Constraints) Concatenation:")
    # display(ipd.Audio(y_algo3, rate=sr))
    # print(f"3. Algo 1 (Manual Guided - No Constraints) Concatenation:")
    # display(ipd.Audio(y_algo1, rate=sr))
    
    return y_orig, y_algos
```
