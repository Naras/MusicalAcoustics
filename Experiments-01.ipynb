```python
def get_scale(num: int):
    if 1 <= num <= 72:
        num -= 1
        # chakra = chakras[num // 6]
        ma = ['M₁', 'M₂'][num // 36]
        by6, by6chakra_ga, by3chakra_ri = num % 6, (num // 6) % 6, (num // 6) % 6
        ni = ['N₁', 'N₂', 'N₃', 'N₂', 'N₃', 'N₃'][by6]
        # da = 'D₁' if by < 3 else 'D₂' if by3 < 6 else 'D₃'
        da = ['D₁', 'D₁', 'D₁', 'D₂', 'D₂', 'D₃'][by6]
        ga = ['G₁', 'G₂', 'G₃', 'G₂', 'G₃', 'G₃'][by6chakra_ga]
        # ri = 'R₁'if by3chakra_ri < 3 else 'R₂' if by3chakra_ri < 6 else 'R₃'
        ri = ['R₁', 'R₁', 'R₁', 'R₂', 'R₂', 'R₃'][by6chakra_ga]
        return f'S {ri} {ga} {ma} P {da} {ni}'
    else: raise ValueError(f'raga no. {num} invalid')
```

```python
import numpy as np
import ipywidgets as widgets
from IPython.display import display, Audio, clear_output
import random
import io
import base64
from scipy.io import wavfile
from scipy.signal import iirpeak, lfilter

# --- DSP Synthesis Engine ---

# 1. Piano Synthesizer
def synth_piano(freq, duration, sr=22050):
    t = np.linspace(0, duration, int(sr * duration), endpoint=False)
    signal = np.zeros_like(t)
    num_harmonics = 8
    B = 0.00015  # Inharmonicity coefficient (stiff strings)
    
    # Generate harmonics with frequency-dependent decay rates
    for k in range(1, num_harmonics + 1):
        f_k = k * freq * np.sqrt(1 + B * k**2)
        if f_k > sr / 2:
            break
        decay_rate = 0.8 + 0.4 * k
        amp = (1.0 / (k ** 1.2)) * np.exp(-decay_rate * t)
        signal += amp * np.sin(2 * np.pi * f_k * t)
        
    # Add hammer strike noise at the attack
    strike_len = int(0.015 * sr)
    if strike_len < len(t):
        noise = np.random.normal(0, 1, strike_len)
        filtered_noise = np.zeros_like(noise)
        for i in range(1, len(noise)):
            filtered_noise[i] = noise[i] + 0.7 * filtered_noise[i-1]
        noise_env = np.exp(-np.linspace(0, 5, strike_len))
        filtered_noise = 0.08 * filtered_noise / np.max(np.abs(filtered_noise)) * noise_env
        signal[:strike_len] += filtered_noise
        
    # Attack envelope
    attack_samples = int(0.005 * sr)
    envelope = np.ones_like(t)
    if attack_samples > 0:
        envelope[:attack_samples] = np.linspace(0, 1, attack_samples)
    
    # Smooth release at the end
    release_samples = int(0.15 * sr)
    if release_samples < len(t):
        envelope[-release_samples:] *= np.linspace(1, 0, release_samples)
        
    signal = signal * envelope
    max_val = np.max(np.abs(signal))
    if max_val > 0:
        signal = signal / max_val * 0.8
    return signal

# 2. Violin Synthesizer
def synth_violin(freq, duration, sr=22050):
    t = np.linspace(0, duration, int(sr * duration), endpoint=False)
    vib_freq = 6.0
    vib_depth = 0.008
    inst_phase = 2 * np.pi * freq * (t - (vib_depth / (2 * np.pi * vib_freq)) * np.cos(2 * np.pi * vib_freq * t))
    
    signal = np.zeros_like(t)
    num_harmonics = 15
    for k in range(1, num_harmonics + 1):
        f_k = k * freq
        if f_k > sr / 2:
            break
        resonance_factor = 1.0
        if f_k < 400:
            resonance_factor *= 1.2
        elif 800 < f_k < 1500:
            resonance_factor *= 1.4
        elif f_k > 3000:
            resonance_factor *= 0.3
            
        amp = (1.0 / k) * resonance_factor
        signal += amp * np.sin(k * inst_phase)
        
    # Subtle bow scraping noise
    noise = np.random.normal(0, 0.015, len(t))
    filtered_noise = np.zeros_like(noise)
    for i in range(1, len(noise)):
        filtered_noise[i] = noise[i] + 0.8 * filtered_noise[i-1]
    signal += 0.03 * filtered_noise
    
    # Bowing ADSR envelope
    attack_samples = int(0.12 * sr)
    release_samples = int(0.15 * sr)
    envelope = np.ones_like(t)
    if attack_samples > 0:
        envelope[:attack_samples] = np.linspace(0, 1, attack_samples)
    if release_samples < len(t):
        envelope[-release_samples:] = np.linspace(1, 0, release_samples)
        
    signal = signal * envelope
    max_val = np.max(np.abs(signal))
    if max_val > 0:
        signal = signal / max_val * 0.7
    return signal

# 3. Voice Synthesizer Helper: Glottal Wave
def get_glottal_source(freq, duration, vibrato_freq, vibrato_depth, sr=22050):
    t = np.linspace(0, duration, int(sr * duration), endpoint=False)
    inst_phase = 2 * np.pi * freq * (t - (vibrato_depth / (2 * np.pi * vibrato_freq)) * np.cos(2 * np.pi * vibrato_freq * t))
    
    source = np.zeros_like(t)
    max_harm = min(30, int(sr / (2 * freq)))
    for k in range(1, max_harm + 1):
        amp = 1.0 / (k ** 1.8)
        source += amp * np.sin(k * inst_phase)
        
    # High-frequency aspiration noise
    noise = np.random.normal(0, 0.005, len(t))
    source += noise
    return source

# 4. Voice Formant Filter
def apply_formant_filter(source, fc, bw, sr, gain):
    b, a = iirpeak(fc, fc/bw, fs=sr)
    filtered = lfilter(b, a, source)
    return filtered * gain

# 5. Human Voice Synthesizer
def synth_voice(freq, duration, voice_type="male", sr=22050):
    if voice_type == "male":
        vib_freq = 5.2
        vib_depth = 0.006
        formants = [
            (730, 50, 1.0),
            (1090, 60, 0.5),
            (2440, 100, 0.15),
            (3400, 120, 0.08)
        ]
    else:
        vib_freq = 5.8
        vib_depth = 0.007
        formants = [
            (850, 60, 1.0),
            (1220, 70, 0.55),
            (2810, 110, 0.22),
            (3750, 130, 0.12)
        ]
        
    source = get_glottal_source(freq, duration, vib_freq, vib_depth, sr)
    output = np.zeros_like(source)
    for fc, bw, gain in formants:
        output += apply_formant_filter(source, fc, bw, sr, gain)
        
    t = np.linspace(0, duration, len(source), endpoint=False)
    attack_samples = int(0.08 * sr)
    release_samples = int(0.12 * sr)
    envelope = np.ones_like(t)
    if attack_samples > 0:
        envelope[:attack_samples] = np.linspace(0, 1, attack_samples)
    if release_samples < len(t):
        envelope[-release_samples:] = np.linspace(1, 0, release_samples)
        
    output = output * envelope
    max_val = np.max(np.abs(output))
    if max_val > 0:
        output = output / max_val * 0.7
    return output

# 6. Sitar Synthesizer
def synth_sitar(freq, duration, sr=22050):
    t = np.linspace(0, duration, int(sr * duration), endpoint=False)
    signal = np.zeros_like(t)
    num_harmonics = 18
    decay_base = 2.5
    
    for k in range(1, num_harmonics + 1):
        f_k = k * freq
        if f_k > sr / 2:
            break
        amp_factor = 1.0 / (k ** 1.1)
        decay_rate = decay_base * (1.0 + 0.1 * k)
        amp = amp_factor * np.exp(-decay_rate * t)
        
        # Sitar meend
        glide = 0.05 * np.exp(-40 * t)
        phase = 2 * np.pi * f_k * (t + glide * t)
        signal += amp * np.sin(phase)
        
    max_sig = np.max(np.abs(signal))
    if max_sig > 0:
        norm_sig = signal / max_sig
        buzz_sig = norm_sig + 0.25 * (norm_sig ** 2) - 0.1 * (norm_sig ** 3)
        
        sym_sig = np.zeros_like(t)
        sym_freqs = [freq, freq * 2, freq * 1.5]
        for sf in sym_freqs:
            if sf < sr / 2:
                sym_sig += np.exp(-0.8 * t) * np.sin(2 * np.pi * sf * t)
        
        signal = buzz_sig * np.exp(-1.5 * t) + 0.15 * sym_sig

    attack_samples = int(0.005 * sr)
    envelope = np.ones_like(t)
    if attack_samples > 0:
        envelope[:attack_samples] = np.linspace(0, 1, attack_samples)
    
    release_samples = int(0.1 * sr)
    if release_samples < len(t):
        envelope[-release_samples:] *= np.linspace(1, 0, release_samples)
        
    signal = signal * envelope
    max_val = np.max(np.abs(signal))
    if max_val > 0:
        signal = signal / max_val * 0.75
    return signal

# 7. Veena Synthesizer
def synth_veena(freq, duration, sr=22050):
    t = np.linspace(0, duration, int(sr * duration), endpoint=False)
    signal = np.zeros_like(t)
    num_harmonics = 12
    decay_base = 1.6
    
    for k in range(1, num_harmonics + 1):
        f_k = k * freq
        if f_k > sr / 2:
            break
        amp_factor = 1.0 / (k ** 1.3)
        decay_rate = decay_base * (1.0 + 0.08 * k)
        amp = amp_factor * np.exp(-decay_rate * t)
        
        # Veena meend
        glide = 0.04 * np.exp(-25 * t)
        phase = 2 * np.pi * f_k * (t + glide * t)
        signal += amp * np.sin(phase)
        
    max_sig = np.max(np.abs(signal))
    if max_sig > 0:
        norm_sig = signal / max_sig
        buzz_sig = norm_sig + 0.12 * (norm_sig ** 2)
        signal = buzz_sig * np.exp(-1.2 * t)
        
    try:
        b1, a1 = iirpeak(180.0, 180.0/50.0, fs=sr)
        b2, a2 = iirpeak(500.0, 500.0/100.0, fs=sr)
        filtered1 = lfilter(b1, a1, signal)
        filtered2 = lfilter(b2, a2, signal)
        signal = 0.6 * signal + 0.25 * filtered1 + 0.15 * filtered2
    except:
        pass

    attack_samples = int(0.01 * sr)
    envelope = np.ones_like(t)
    if attack_samples > 0:
        envelope[:attack_samples] = np.linspace(0, 1, attack_samples)
        
    release_samples = int(0.15 * sr)
    if release_samples < len(t):
        envelope[-release_samples:] *= np.linspace(1, 0, release_samples)
        
    signal = signal * envelope
    max_val = np.max(np.abs(signal))
    if max_val > 0:
        signal = signal / max_val * 0.75
    return signal

# Audio routing helper
def generate_audio(inst, freq, dur):
    if inst == 'Piano':
        return synth_piano(freq, dur)
    elif inst == 'Violin':
        return synth_violin(freq, dur)
    elif inst == 'Human Male Voice':
        return synth_voice(freq, dur, 'male')
    elif inst == 'Human Female Voice':
        return synth_voice(freq, dur, 'female')
    elif inst == 'Sitar':
        return synth_sitar(freq, dur)
    elif inst == 'Veena':
        return synth_veena(freq, dur)
    return synth_piano(freq, dur)

# --- End of DSP Synthesis Engine ---

# Local copies to avoid dependency on previous cell execution state
ref_carnatic_16 = ['S', 'R₁', 'R₂', 'R₃', 'G₁', 'G₂', 'G₃', 'M₁', 'M₂', 'P', 'D₁', 'D₂', 'D₃', 'N₁', 'N₂', 'N₃']
ref_western_16 = ['C', 'C#', 'D', 'D#', 'D', 'D#', 'E', 'F', 'F#', 'G', 'G#', 'A', 'A#', 'A', 'A#', 'B']
ref_western_12 = ['C', 'C#', 'D', 'D#', 'E', 'F', 'F#', 'G', 'G#', 'A', 'A#', 'B']

# Kannada and Devanagari translation maps for scale swaras
swara_kannada_map = {
    'S': 'ಸ', 'R₁': 'ರಿ₁', 'R₂': 'ರಿ₂', 'R₃': 'ರಿ₃',
    'G₁': 'ಗ₁', 'G₂': 'ಗ₂', 'G₃': 'ಗ₃',
    'M₁': 'ಮ₁', 'M₂': 'ಮ₂', 'P': 'ಪ',
    'D₁': 'ದ₁', 'D₂': 'ದ₂', 'D₃': 'ದ₃',
    'N₁': 'ನಿ₁', 'N₂': 'ನಿ₂', 'N₃': 'ನಿ₃'
}
swara_devanagari_map = {
    'S': 'स', 'R₁': 'रि₁', 'R₂': 'रि₂', 'R₃': 'रि₃',
    'G₁': 'ग₁', 'G₂': 'ग₂', 'G₃': 'ग₃',
    'M₁': 'म₁', 'M₂': 'म₂', 'P': 'प',
    'D₁': 'द₁', 'D₂': 'द₂', 'D₃': 'द₃',
    'N₁': 'नि₁', 'N₂': 'नि₂', 'N₃': 'नि₃'
}

def get_swara_frequency(swara, octave):
    try:
        idx = ref_carnatic_16.index(swara)
        western_note = ref_western_16[idx]
        note_index = ref_western_12.index(western_note)
        midi_note = 12 * (octave + 1) + note_index
        return 440.0 * (2.0 ** ((midi_note - 69) / 12.0))
    except Exception as e:
        return 261.63  # Fallback to C4

# Synthesizer and Control widgets (re-added locally to the last cell)
raga_inst_w = widgets.ToggleButtons(
    options=['Piano', 'Violin', 'Human Male Voice', 'Human Female Voice', 'Sitar', 'Veena'],
    description='Instrument:',
    disabled=False,
    button_style='info',
)

raga_octave_w = widgets.ToggleButtons(
    options=[3, 4, 5],
    value=4,
    description='Octave:',
    disabled=False,
    button_style='',
)

raga_dur_w = widgets.FloatSlider(
    value=1.2,
    min=0.2,
    max=3.0,
    step=0.1,
    description='Duration (s):',
    continuous_update=False
)

# Native Python hold controls
raga_hold_mode_w = widgets.Checkbox(
    value=False,
    description='Hold Note (Loop Mode)',
    indent=False,
    layout=widgets.Layout(width='180px')
)
btn_stop_note = widgets.Button(
    description="STOP",
    button_style='danger',
    layout=widgets.Layout(width='80px')
)

# Raga selection controls
raga_num_w = widgets.IntSlider(value=29, min=1, max=72, step=1, description="Raga No:")
raga_num_text_w = widgets.BoundedIntText(value=29, min=1, max=72, step=1, description="Enter No:")

# Bi-directional sync via jslink
widgets.jslink((raga_num_w, 'value'), (raga_num_text_w, 'value'))

details_out = widgets.HTML()
swara_buttons_container = widgets.VBox(layout=widgets.Layout(margin='5px 0'))
raga_output_w = widgets.Output()

# Phrase length slider & play random button
raga_phrase_len_w = widgets.IntSlider(
    value=8,
    min=4,
    max=16,
    step=1,
    description='Phrase Length:',
    layout=widgets.Layout(width='300px')
)
btn_play_random = widgets.Button(
    description="Play Random Phrase",
    button_style='danger',
    layout=widgets.Layout(width='170px')
)

def update_raga_display(*args):
    num = raga_num_w.value
    raga = ragaByNumber.get(num, {})
    scale_str = get_scale(num)
    scale_notes = scale_str.split()
    
    kannada_scale = [swara_kannada_map.get(s, s) for s in scale_notes]
    devanagari_scale = [swara_devanagari_map.get(s, s) for s in scale_notes]
    
    western_notes = []
    for s in scale_notes:
        try:
            idx = ref_carnatic_16.index(s)
            western_notes.append(ref_western_16[idx])
        except:
            western_notes.append(s)
            
    details_html = f"""
    <div style="background-color: #1e3c72; color: white; padding: 15px; border-radius: 10px; border: 1px solid #122c5a; margin-top: 10px; max-width: 580px;">
        <h3 style="margin-top: 0; color: #ffca28; border-bottom: 1px solid rgba(255,255,255,0.2); padding-bottom: 5px;">Melakarta Raga #{num}: {raga.get('IAST', '')}</h3>
        <table style="width: 100%; border-collapse: collapse; margin-bottom: 10px;">
            <tr>
                <td style="font-weight: bold; width: 35%; padding: 4px 0; color: #b2dfdb;">IAST Name:</td>
                <td style="padding: 4px 0; font-size: 1.1em;">{raga.get('IAST', '')}</td>
            </tr>
            <tr>
                <td style="font-weight: bold; padding: 4px 0; color: #b2dfdb;">Devanagari:</td>
                <td style="padding: 4px 0; font-size: 1.1em;">{raga.get('devanagari', '')}</td>
            </tr>
            <tr>
                <td style="font-weight: bold; padding: 4px 0; color: #b2dfdb;">Kannada (ಕನ್ನಡ):</td>
                <td style="padding: 4px 0; font-size: 1.1em;">{raga.get('kannada', '')}</td>
            </tr>
            <tr>
                <td style="font-weight: bold; padding: 4px 0; color: #e0f2f1;">Chakra (ಚಕ್ರ / चक्र):</td>
                <td style="padding: 4px 0; font-size: 1.1em; font-weight: 500; color: #ffca28;">{raga.get('chakra', '')}</td>
            </tr>
        </table>
        
        <div style="margin-top: 15px; background: rgba(255,255,255,0.1); padding: 12px; border-radius: 8px; border-left: 4px solid #ffeb3b;">
            <h4 style="margin: 0 0 8px 0; color: #ffeb3b; font-size: 1.1em;">Scale Swaras (get_scale):</h4>
            <div style="margin-bottom: 5px;"><strong>English (IAST):</strong> <span style="font-size: 1.25em; letter-spacing: 2px; color: #ffffff;">{' '.join(scale_notes)}</span></div>
            <div style="margin-bottom: 5px;"><strong>Kannada (ಕನ್ನಡ):</strong> <span style="font-size: 1.25em; letter-spacing: 2px; color: #b2dfdb;">{' '.join(kannada_scale)}</span></div>
            <div style="margin-bottom: 5px;"><strong>Devanagari (देवनागरी):</strong> <span style="font-size: 1.25em; letter-spacing: 2px; color: #b2dfdb;">{' '.join(devanagari_scale)}</span></div>
            <div><strong>Western notes:</strong> <span style="font-size: 1.1em; letter-spacing: 1.5px; color: #ffcc80;">{' '.join(western_notes)}</span></div>
        </div>
    </div>
    """
    details_out.value = details_html

def play_single_swara(swara, oct_val):
    with raga_output_w:
        clear_output(wait=True)
        inst = raga_inst_w.value
        dur = raga_dur_w.value
        freq = get_swara_frequency(swara, oct_val)
        
        loop_mode = raga_hold_mode_w.value
        loop_attr = "loop" if loop_mode else ""
        
        sig = generate_audio(inst, freq, dur)
        
        # Write to memory buffer
        buf = io.BytesIO()
        wavfile.write(buf, 22050, (sig * 32767).astype(np.int16))
        b64_data = base64.b64encode(buf.getvalue()).decode('utf-8')
        
        mode_label = "Loop Mode" if loop_mode else "Single Play"
        print_str = f"Playing note: {swara} ({freq:.2f} Hz) on {inst} (Octave {oct_val}) | {mode_label}"
        
        audio_html = f"""
        <div style="margin-top: 5px;">
            <p style="font-weight: bold; color: #2a5298; margin: 0 0 5px 0;">{print_str}</p>
            <audio autoplay controls {loop_attr} style="width: 100%; max-width: 400px;">
                <source src="data:audio/wav;base64,{b64_data}" type="audio/wav">
            </audio>
        </div>
        """
        display(widgets.HTML(audio_html))

def stop_note_action(b):
    with raga_output_w:
        clear_output(wait=True)
        print("Note stopped.")

btn_stop_note.on_click(stop_note_action)

def update_swara_buttons(*args):
    num = raga_num_w.value
    scale_str = get_scale(num)
    scale_notes = scale_str.split()
    
    octaves_list = [3, 4, 5]
    rows = []
    
    for idx, oct_val in enumerate(octaves_list):
        notes_list = [(s, oct_val) for s in scale_notes]
        if idx == len(octaves_list) - 1:
            notes_list.append(('S', oct_val + 1))
            
        row_buttons = []
        for s, o in notes_list:
            is_high = o > oct_val
            suffix = "⁺" if is_high else ""
            btn = widgets.Button(
                description=f"{s}{suffix}", 
                layout=widgets.Layout(width='60px', height='40px', margin='2px')
            )
            btn.style.button_color = '#e1f5fe' if is_high else '#f5f5f5'
            btn.style.text_color = '#0288d1' if is_high else 'black'
            btn.style.font_weight = 'bold'
            
            btn.on_click(lambda b, sw=s, oc=o: play_single_swara(sw, oc))
            row_buttons.append(btn)
            
        label = widgets.HTML(
            f"<div style='font-size:0.9em; font-weight:bold; color:#555; width:70px; margin-top:8px;'>Octave {oct_val}:</div>"
        )
        rows.append(widgets.HBox([label] + row_buttons))
        
    swara_buttons_container.children = rows

raga_num_w.observe(update_raga_display, names='value')
raga_num_w.observe(update_swara_buttons, names='value')

# Initialize
update_raga_display()
update_swara_buttons()

def play_raga_scale_action(order_type):
    with raga_output_w:
        clear_output(wait=True)
        raga_num = raga_num_w.value
        raga_details = ragaByNumber.get(raga_num, {})
        raga_name = raga_details.get('IAST', '')
        
        scale_str = get_scale(raga_num)
        swaras = scale_str.split()
        
        inst = raga_inst_w.value
        octave = raga_octave_w.value
        dur = raga_dur_w.value
        
        aro = [(s, octave) for s in swaras]
        extra = [('S', octave + 1)]
        avar = [(s, octave) for s in reversed(swaras)]
        
        sr = 22050
        
        if order_type == 'Arohana':
            seq = aro + extra
            label = "Arohana (Ascending)"
            signals = [generate_audio(inst, get_swara_frequency(s, oct_val), dur) for s, oct_val in seq]
            combined = np.concatenate(signals)
            seq_display = [f"{s}" if oct_val == octave else f"{s}⁺" for s, oct_val in seq]
        elif order_type == 'Avarohana':
            seq = extra + avar
            label = "Avarohana (Descending)"
            signals = [generate_audio(inst, get_swara_frequency(s, oct_val), dur) for s, oct_val in seq]
            combined = np.concatenate(signals)
            seq_display = [f"{s}" if oct_val == octave else f"{s}⁺" for s, oct_val in seq]
        elif order_type == 'Arohana-Avarohana':
            seq = aro + extra + extra + avar
            label = "Arohana-Avarohana (Combination)"
            signals = [generate_audio(inst, get_swara_frequency(s, oct_val), dur) for s, oct_val in seq]
            combined = np.concatenate(signals)
            seq_display = [f"{s}" if oct_val == octave else f"{s}⁺" for s, oct_val in seq]
        elif order_type == 'FullSequence':
            seq_aro = aro + extra
            seq_avar = extra + avar
            seq_comb = aro + extra + extra + avar
            
            silence = np.zeros(int(sr * 0.4))
            
            sig_aro = np.concatenate([generate_audio(inst, get_swara_frequency(s, oct_val), dur) for s, oct_val in seq_aro])
            sig_avar = np.concatenate([generate_audio(inst, get_swara_frequency(s, oct_val), dur) for s, oct_val in seq_avar])
            sig_comb = np.concatenate([generate_audio(inst, get_swara_frequency(s, oct_val), dur) for s, oct_val in seq_comb])
            
            combined = np.concatenate([sig_aro, silence, sig_avar, silence, sig_comb])
            label = "Full Scale Practice (Arohana → Avarohana → Combination)"
            seq_display = ["Arohana", "Avarohana", "Arohana-Avarohana"]
            
        # Write to memory buffer
        buf = io.BytesIO()
        wavfile.write(buf, sr, (combined * 32767).astype(np.int16))
        b64_data = base64.b64encode(buf.getvalue()).decode('utf-8')
        
        print_str = f"🎵 Playing: Raga #{raga_num} ({raga_name}) - {label}"
        audio_html = f"""
        <div style="margin-top: 5px;">
            <p style="font-weight: bold; color: #2a5298; margin: 0 0 5px 0;">{print_str}</p>
            <audio autoplay controls style="width: 100%; max-width: 400px;">
                <source src="data:audio/wav;base64,{b64_data}" type="audio/wav">
            </audio>
        </div>
        """
        display(widgets.HTML(audio_html))

def play_random_phrase(b):
    with raga_output_w:
        clear_output(wait=True)
        num = raga_num_w.value
        raga_details = ragaByNumber.get(num, {})
        raga_name = raga_details.get('IAST', '')
        
        scale_str = get_scale(num)
        swaras = scale_str.split()
        
        inst = raga_inst_w.value
        octave = raga_octave_w.value
        dur = raga_dur_w.value
        length = raga_phrase_len_w.value
        
        # Build note pool from base octave swaras + high S
        pool = [(s, octave) for s in swaras] + [('S', octave + 1)]
        
        seq = [random.choice(pool) for _ in range(length)]
        
        sr = 22050
        signals = []
        seq_display = []
        for s, oct_val in seq:
            freq = get_swara_frequency(s, oct_val)
            sig = generate_audio(inst, freq, dur)
            signals.append(sig)
            
            suffix = "⁺" if oct_val > octave else ""
            seq_display.append(f"{s}{suffix}")
            
        combined = np.concatenate(signals)
        
        # Write to memory buffer
        buf = io.BytesIO()
        wavfile.write(buf, sr, (combined * 32767).astype(np.int16))
        b64_data = base64.b64encode(buf.getvalue()).decode('utf-8')
        
        print_str = f"🎲 Random phrase in Raga #{num} ({raga_name}): {' → '.join(seq_display)}"
        audio_html = f"""
        <div style="margin-top: 5px;">
            <p style="font-weight: bold; color: #2a5298; margin: 0 0 5px 0;">{print_str}</p>
            <audio autoplay controls style="width: 100%; max-width: 400px;">
                <source src="data:audio/wav;base64,{b64_data}" type="audio/wav">
            </audio>
        </div>
        """
        display(widgets.HTML(audio_html))

btn_raga_aro = widgets.Button(description="Arohana", button_style='primary', layout=widgets.Layout(width='120px'))
btn_raga_avar = widgets.Button(description="Avarohana", button_style='primary', layout=widgets.Layout(width='120px'))
btn_raga_comb = widgets.Button(description="Combination", button_style='success', layout=widgets.Layout(width='120px'))
btn_raga_full = widgets.Button(description="Play Full Sequence", button_style='warning', layout=widgets.Layout(width='160px'))

btn_raga_aro.on_click(lambda b: play_raga_scale_action('Arohana'))
btn_raga_avar.on_click(lambda b: play_raga_scale_action('Avarohana'))
btn_raga_comb.on_click(lambda b: play_raga_scale_action('Arohana-Avarohana'))
btn_raga_full.on_click(lambda b: play_raga_scale_action('FullSequence'))
btn_play_random.on_click(play_random_phrase)

raga_ui = widgets.VBox([
    widgets.HTML("<h2>🎶 Melakarta Raga Scale Synthesizer 🎶</h2>"),
    
    # Synthesizer Controls (re-added locally to the last cell)
    widgets.VBox([
        widgets.HTML("<h4>Instrument & Synthesis Options:</h4>"),
        raga_inst_w,
        raga_octave_w,
        raga_dur_w
    ], layout=widgets.Layout(margin='5px 0', border='1px solid #ccc', padding='10px', border_radius='8px')),
    
    # Raga selectors
    widgets.HBox([raga_num_w, raga_num_text_w]),
    
    # Details card
    details_out,
    
    # Play Individual Notes
    widgets.VBox([
        widgets.HTML("<h4>Play Individual Scale Swaras (Click to play):</h4>"),
        widgets.HBox([raga_hold_mode_w, btn_stop_note], layout=widgets.Layout(align_items='center', margin='0 0 10px 0')),
        swara_buttons_container
    ], layout=widgets.Layout(margin='5px 0', border='1px solid #ccc', padding='10px', border_radius='8px')),
    
    # Scale Playback Buttons
    widgets.VBox([
        widgets.HTML("<h4>Scale Playback:</h4>"),
        widgets.HBox([btn_raga_aro, btn_raga_avar, btn_raga_comb, btn_raga_full])
    ], layout=widgets.Layout(margin='5px 0')),
    
    # Random Phrase Generator
    widgets.VBox([
        widgets.HTML("<h4>Random Sequence Generator:</h4>"),
        widgets.HBox([raga_phrase_len_w, btn_play_random])
    ], layout=widgets.Layout(margin='5px 0', border='1px solid #ccc', padding='10px', border_radius='8px')),
    
    # Audio output
    widgets.HTML("<h4 style='margin-top: 15px;'>Raga Audio Output:</h4>"),
    raga_output_w
], layout=widgets.Layout(
    padding='20px', 
    max_width='600px', 
    border='2px solid #2a5298', 
    border_radius='12px', 
    background_color='#f0f4f8'
))

display(raga_ui)
```
