```python
def get_scale(num: int):
    if 1 <= num <= 72:
        num -= 1
        # chakra = chakras[num // 6]
        ma = ['M₁', 'M₂'][num // 36]
        by6, by6chakra_ga, by3chakra_ri = num % 6, (num // 6) % 6, (num // 6) % 6
        ni = ['N₁', 'N₂', 'N₃', 'N₂', 'N₃', 'N₃'][by6]
        # da = 'D₁' if by < 3 else 'D₂' if by3 < 6 else 'D₃'
        da = ['D₁', 'D₁', 'D₁', 'D₂', 'D₂', 'D₃'][by6]
        ga = ['G₁', 'G₂', 'G₃', 'G₂', 'G₃', 'G₃'][by6chakra_ga]
        # ri = 'R₁'if by3chakra_ri < 3 else 'R₂' if by3chakra_ri < 6 else 'R₃'
        ri = ['R₁', 'R₁', 'R₁', 'R₂', 'R₂', 'R₃'][by6chakra_ga]
        return f'S {ri} {ga} {ma} P {da} {ni}'
    else: raise ValueError(f'raga no. {num} invalid')
```

```python
import numpy as np
import ipywidgets as widgets
from IPython.display import display, Audio, clear_output
import random
import io
import base64
import os
import librosa
import scipy.signal
from scipy.io import wavfile
from scipy.signal import iirpeak, lfilter

# --- DSP Synthesis Engine ---

# Load vocal audio sample file globally if available to support Voice instruments
audio_path = '/Users/narasimhanm.g./Desktop/MusicAlbums/0Carnatic/0AmmaLessons/01saraLe-varasegaLu.MP3'
y_vocal = None
if os.path.exists(audio_path):
    try:
        y_vocal, _ = librosa.load(audio_path, sr=22050)
        print("Voice audio sample file loaded successfully.")
    except Exception as e:
        print("Warning: Could not load voice audio file. Vocal options will fallback to synthesized piano. Error:", e)
else:
    print("Warning: Voice audio file path not found. Vocal options will fallback to synthesized piano.")


# General signal concatenation function with crossfade to ensure smooth Arohana/Avarohana playback
def concatenate_signals(signals, sr, crossfade_ms=15):
    if not signals:
        return np.array([])
    fade_len = int((crossfade_ms * sr) / 1000)
    y_concat = signals[0].copy()
    for next_slice in signals[1:]:
        if fade_len > 0 and len(y_concat) >= fade_len and len(next_slice) >= fade_len:
            fade_out = np.linspace(1.0, 0.0, fade_len)
            fade_in = np.linspace(0.0, 1.0, fade_len)
            y_concat[-fade_len:] = y_concat[-fade_len:] * fade_out
            faded_next = next_slice.copy()
            faded_next[:fade_len] = faded_next[:fade_len] * fade_in
            overlap = y_concat[-fade_len:] + faded_next[:fade_len]
            y_concat = np.concatenate([y_concat[:-fade_len], overlap, faded_next[fade_len:]])
        else:
            y_concat = np.concatenate([y_concat, next_slice])
    return y_concat

# Formant-preserving pitch shifting using Cepstral compensation & filtering (Option 1 V3)
def pitch_shift_formant_preserved_v3(y_sig, sr=22050, n_steps=12, n_fft=2048, hop_length=512,
                                      lifter_cutoff=18, ref_segment=None):
    """
    Formant-preserved pitch shifting (Option 1 V3):
    - Cepstral spectral envelope extraction and ratio compensation filter.
    - High-frequency shelf de-emphasis above 3.5 kHz (eliminates squeakiness).
    - Low-frequency subharmonic neutralization (eliminates double-voice/ghost octave).
    - Direction-aware high-pass filter (preserves low-octave bass while cleaning high-octave rumble).
    - Dynamic RMS level matching.
    """
    if abs(n_steps) < 0.1:
        return y_sig
    y_shifted = librosa.effects.pitch_shift(y_sig, sr=sr, n_steps=n_steps, n_fft=n_fft, hop_length=hop_length)

    def get_envelope(sig):
        D = librosa.stft(sig, n_fft=n_fft, hop_length=hop_length)
        lm = np.log(np.abs(D) + 1e-8)
        c = np.fft.ifft(lm, axis=0)
        lif = np.zeros_like(c); lif[:lifter_cutoff] = c[:lifter_cutoff]; lif[-lifter_cutoff:] = c[-lifter_cutoff:]
        return np.exp(np.real(np.fft.fft(lif, axis=0))), D

    env_orig, _ = get_envelope(y_sig)
    env_shift, D_shift = get_envelope(y_shifted)

    mt = min(env_orig.shape[1], env_shift.shape[1], D_shift.shape[1])
    env_orig = env_orig[:, :mt]; env_shift = env_shift[:, :mt]; D_shift = D_shift[:, :mt]

    filt = np.clip(env_orig / (env_shift + 1e-6), 0.2, 5.0)
    freqs = librosa.fft_frequencies(sr=sr, n_fft=n_fft)
    
    # Low-frequency subharmonic neutralisation (anti double-voice for upward shifts)
    if n_steps > 0:
        lo_bin = np.searchsorted(freqs, 350.0)
        for i in range(lo_bin):
            w = float(i) / lo_bin
            filt[i, :] = 1.0 * (1 - w) + filt[i, :] * w

    # High-frequency shelf de-emphasis above 3.5 kHz (anti squeakiness)
    hi_bin = np.searchsorted(freqs, 3500.0)
    n_bins = len(freqs)
    for i in range(hi_bin, n_bins):
        shelf = 1.0 - 0.4 * min(1.0, (i - hi_bin) / max(1, n_bins - hi_bin))
        filt[i, :] *= shelf

    D_comp = D_shift * filt
    _, y_out = scipy.signal.istft(D_comp, fs=sr, nperseg=n_fft, noverlap=n_fft - hop_length)

    # Direction-appropriate high-pass cutoff: 60 Hz for downward shift (Octave 3 bass), 250 Hz for upward shift (Octave 5)
    hp_cutoff = 60 if n_steps < 0 else 250
    sos = scipy.signal.butter(4, hp_cutoff, 'hp', fs=sr, output='sos')
    y_out = scipy.signal.sosfilt(sos, y_out)

    if len(y_out) > len(y_shifted): y_out = y_out[:len(y_shifted)]
    elif len(y_out) < len(y_shifted): y_out = np.pad(y_out, (0, len(y_shifted) - len(y_out)))

    # RMS normalise to reference segment level
    ref = ref_segment if ref_segment is not None else y_sig
    ref_rms = np.sqrt(np.mean(ref**2)) + 1e-9
    out_rms = np.sqrt(np.mean(y_out**2)) + 1e-9
    return y_out * (ref_rms / out_rms)

# Helper to load and pitch-shift vocal extracts using Formant-Preserved V3
# Helper to load and pitch-shift vocal extracts using Formant-Preserved V3
def extract_manual_vocal_swara(voice, swara_symbol, target_freq, dur, octave):
    mapping = {
        'S':  ('ascent_1', 'sa'),
        'R₁': ('ascent_1', 'ri1'),
        'R₂': ('ascent_1', 'ri1'),
        'R₃': ('ascent_1', 'ri1'),
        'G₁': ('ascent_1', 'ga3'),
        'G₂': ('ascent_1', 'ga3'),
        'G₃': ('ascent_1', 'ga3'),
        'M₁': ('ascent_1', 'ma1'),
        'M₂': ('ascent_1', 'ma1'),
        'P':  ('ascent_2', 'pa'),
        'D₁': ('ascent_2', 'da1'),
        'D₂': ('ascent_2', 'da1'),
        'D₃': ('ascent_2', 'da1'),
        'N₁': ('ascent_2', 'ni3'),
        'N₂': ('ascent_2', 'ni3'),
        'N₃': ('ascent_2', 'ni3'),
    }
    
    mapping_high = {
        # 'S':  ('sa-ri1-ga3-ma1-pa-da1-ni3-sa2', 'sa2'),
        'R₁': ('da1-ni3-sa2-ri1', 'ri1'),
        'R₂': ('da1-ni3-sa2-ri1', 'ri1'),
        'R₃': ('da1-ni3-sa2-ri1', 'ri1'),
        'G₁': ('da1-ni3-sa2-ri1-ga3-ri1-sa-ri1', 'ga3'),
        'G₂': ('da1-ni3-sa2-ri1-ga3-ri1-sa-ri1', 'ga3'),
        'G₃': ('da1-ni3-sa2-ri1-ga3-ri1-sa-ri1', 'ga3'),
        'M₁': ('da1-ni3-sa2-ri1-ga3-ma1-pa-ma1', 'ma1'),
        'M₂': ('da1-ni3-sa2-ri1-ga3-ma1-pa-ma1', 'ma1'),
        'P':  ('da1-ni3-sa2-ri1-ga3-ma1-pa-ma1', 'pa'),
    }

    mapping_low = {
        # 'S':  ('sa-ni1-da1-pa-ma1-ga3-ri1-sa2', 'sa2'),
        'D₁': ('ga3-ri1-sa2-ni3-da1-ni3-sa-ni3', 'da1'),
        'D₂': ('ga3-ri1-sa2-ni3-da1-ni3-sa-ni3', 'da1'),
        'N₁': ('ga3-ri1-sa2-ni3-da1-ni3-sa-ni3', 'ni3'),
        'N₂': ('ga3-ri1-sa2-ni3-da1-ni3-sa-ni3', 'ni3'),
        'N₃': ('ga3-ri1-sa2-ni3-da1-ni3-sa-ni3', 'ni3'),
        'P':  ('ga3-ri1-sa2-ni3-da1-pa-da1-ni3', 'pa'),
    }
    # print(f'inputs - voice {voice} swara {swara_symbol} freq {target_freq} octave {octave}')

    # Select the target recorded note from middle octave .. this will the defualt base for any shifts later
    if swara_symbol == 'S':
        if target_freq > 300.0 or (190.0 < target_freq < 280.0): phrase_key, note_key = ('ascent_2', 'sa2')
        else: phrase_key, note_key = ('ascent_1', 'sa')
    else: phrase_key, note_key = mapping.get(swara_symbol, ('ascent_1', 'sa'))    
        
    if octave == 5: 
        if swara_symbol in mapping_high: phrase_key, note_key = mapping_high[swara_symbol] #.get(swara_symbol, ('da1-ni3-sa2-ri1', 'sa'))
    elif octave == 3: 
        if swara_symbol in mapping_low: phrase_key, note_key = mapping_low[swara_symbol] #.get(swara_symbol, ('ga3-ri1-sa2-ni3', 'sa'))
    # print(f'debug1 new maps teacher voice... phrase {phrase_key} note {note_key}')
    
    # Reference base tonic: Teacher naturally sings at ~145 Hz, Student naturally sings at ~72.5-90 Hz
    f_sa_ref = 145.0 # if voice == 'teacher' else 72.5
    semitones_map = {
        'sa': 0, 'ri1': 1, 'ga3': 4, 'ma1': 5,
        'pa': 7, 'da1': 8, 'ni3': 11, 'sa2': 12
    }
    
    base_semitone = semitones_map.get(note_key, 0)
    f_base = f_sa_ref * (2.0 ** (base_semitone / 12.0))
    
    # print(f'debug2 new maps {voice} voice... octave {octave} phrase {phrase_key} note {note_key}')

    if octave == 4: time_tuple = manual_selections[voice][phrase_key]["notes"].get(note_key, (3, 44, 3, 45.75))
    elif octave == 5: 
        if swara_symbol in mapping_high: time_tuple = octave_high[voice][phrase_key]["notes"].get(note_key, (20, 55.7, 20, 57))
        else: time_tuple = manual_selections[voice][phrase_key]["notes"].get(note_key, (3, 44, 3, 45.75))
    elif octave == 3: 
        if swara_symbol in mapping_low: time_tuple = octave_low[voice][phrase_key]["notes"].get(note_key, (43, 8.9, 43, 10))
        else: time_tuple = manual_selections[voice][phrase_key]["notes"].get(note_key, (3, 44, 3, 45.75))
    
    # special case high sa .. in octave 5        
    if swara_symbol == 'S':
        if target_freq > 300.0 or (190.0 < target_freq < 280.0): 
            phrase_key, note_key = ('ascent_2', 'sa2')
            time_tuple = manual_selections[voice][phrase_key]["notes"].get(note_key, (3, 44, 3, 45.75))
            
    t_start = time_tuple[0] * 60 + time_tuple[1]
    t_end = time_tuple[2] * 60 + time_tuple[3]
    
    idx_start = int(t_start * 22050)
    idx_end = int(t_end * 22050)
    
    if y_vocal is None or len(y_vocal) == 0: return synth_piano(target_freq * 2.0, dur)
        
    y_slice = y_vocal[idx_start:idx_end].copy()

    # if recorded note exists for the low or high swara, return that. 
    if octave == 5 and swara_symbol in mapping_high: 
        # print(f' play recorded voice {voice} octave:{octave} phrase:{phrase_key} swara:{note_key}')
        return y_slice
    elif octave == 3 and swara_symbol in mapping_low:
        # print(f' play recorded voice {voice} octave:{octave} phrase:{phrase_key} swara:{note_key}')
        return y_slice
    
    # if octave == 4:print(f' play synthesized formant preserved octave:{octave} phrase:{phrase_key} swara:{note_key}')
    
    # Use Formant-Preserved V3 shift for octave movements (+12 st for Octave 5, -12 st for Octave 3)
    total_shift = 12.0 * np.log2(target_freq / f_base)
    if abs(total_shift) > 2.5:
        y_shifted = pitch_shift_formant_preserved_v3(y_slice, 22050, n_steps=total_shift)
    else:
        try:
            y_shifted = librosa.effects.pitch_shift(y_slice, sr=22050, n_steps=total_shift)
        except Exception:
            y_shifted = y_slice
        
    target_samples = int(dur * 22050)
    if len(y_shifted) > target_samples:
        y_shifted = y_shifted[:target_samples]
        fade_len = min(1100, len(y_shifted) // 10)
        if fade_len > 0:
            fade_out = np.linspace(1.0, 0.0, fade_len)
            y_shifted[-fade_len:] = y_shifted[-fade_len:] * fade_out
    elif len(y_shifted) < target_samples:
        padding = np.zeros(target_samples - len(y_shifted))
        y_shifted = np.concatenate([y_shifted, padding])
        
    # Standardize normalization gain to 0.95 for clear vocal reproduction
    max_val = np.max(np.abs(y_shifted))
    if max_val > 0:
        y_shifted = y_shifted / max_val * 0.95
        
    return y_shifted

# 1. Piano Synthesizer
def synth_piano(freq, duration, sr=22050):
    t = np.linspace(0, duration, int(sr * duration), endpoint=False)
    signal = np.zeros_like(t)
    num_harmonics = 8
    B = 0.00015
    for k in range(1, num_harmonics + 1):
        f_k = k * freq * np.sqrt(1 + B * k**2)
        if f_k > sr / 2:
            break
        decay_rate = 0.8 + 0.4 * k
        amp = (1.0 / (k ** 1.2)) * np.exp(-decay_rate * t)
        signal += amp * np.sin(2 * np.pi * f_k * t)
    strike_len = int(0.015 * sr)
    if strike_len < len(t):
        env = np.ones_like(t)
        env[:strike_len] = np.linspace(0.0, 1.0, strike_len)
        signal *= env
    max_val = np.max(np.abs(signal))
    if max_val > 0:
        signal = signal / max_val * 0.75
    return signal

# 2. Violin Synthesizer
def synth_violin(freq, duration, sr=22050):
    t = np.linspace(0, duration, int(sr * duration), endpoint=False)
    signal = np.zeros_like(t)
    num_harmonics = 10
    vibrato_freq = 6.0
    vibrato_depth = 0.01
    phase = 2 * np.pi * freq * (t + vibrato_depth / vibrato_freq * np.sin(2 * np.pi * vibrato_freq * t))
    for k in range(1, num_harmonics + 1):
        amp = 1.0 / k if k <= 3 else 0.5 / (k ** 1.5)
        signal += amp * np.sin(k * phase)
    try:
        b, a = iirpeak(freq, freq / 10.0, fs=sr)
        signal = lfilter(b, a, signal)
    except:
        pass
    env = np.ones_like(t)
    attack_len = int(0.1 * sr)
    decay_len = int(0.15 * sr)
    if attack_len + decay_len < len(t):
        env[:attack_len] = np.linspace(0.0, 1.0, attack_len)
        env[-decay_len:] = np.linspace(1.0, 0.0, decay_len)
        signal *= env
    max_val = np.max(np.abs(signal))
    if max_val > 0:
        signal = signal / max_val * 0.75
    return signal

# 3. Voice Synthesizer
def get_glottal_source(freq, duration, vibrato_freq, vibrato_depth, sr=22050):
    t = np.linspace(0, duration, int(sr * duration), endpoint=False)
    phase = 2 * np.pi * freq * (t + vibrato_depth / vibrato_freq * np.sin(2 * np.pi * vibrato_freq * t))
    glottal = (scipy.signal.sawtooth(phase, 0.9) + 1.0) / 2.0
    return glottal

def apply_formant_filter(source, fc, bw, sr, gain):
    q = fc / bw
    b, a = scipy.signal.iirpeak(fc, q, fs=sr)
    filtered = scipy.signal.lfilter(b, a, source)
    return filtered * gain

def synth_voice(freq, duration, voice_type="male", sr=22050):
    glottal = get_glottal_source(freq, duration, 5.5, 0.008, sr)
    if voice_type == "male":
        formants = [(650, 80, 1.0), (1080, 90, 0.6), (2650, 150, 0.35), (2900, 200, 0.15)]
    else:
        formants = [(800, 90, 1.0), (1150, 100, 0.6), (2800, 160, 0.35), (3200, 220, 0.15)]
    signal = np.zeros_like(glottal)
    for fc, bw, gain in formants:
        signal += apply_formant_filter(glottal, fc, bw, sr, gain)
    env = np.ones_like(glottal)
    attack_len = int(0.08 * sr)
    decay_len = int(0.12 * sr)
    if attack_len + decay_len < len(env):
        env[:attack_len] = np.linspace(0.0, 1.0, attack_len)
        env[-decay_len:] = np.linspace(1.0, 0.0, decay_len)
        signal *= env
    max_val = np.max(np.abs(signal))
    if max_val > 0:
        signal = signal / max_val * 0.75
    return signal

# 4. Sitar Synthesizer
def synth_sitar(freq, duration, sr=22050):
    t = np.linspace(0, duration, int(sr * duration), endpoint=False)
    signal = np.zeros_like(t)
    num_harmonics = 12
    jawari_depth = 0.002
    jawari_freq = 150.0
    for k in range(1, num_harmonics + 1):
        f_k = k * freq
        if f_k > sr / 2:
            break
        mod = 1.0 + jawari_depth * np.sin(2 * np.pi * jawari_freq * t)
        decay_rate = 1.5 + 0.35 * k
        amp = (1.0 / (k ** 1.3)) * np.exp(-decay_rate * t) * mod
        signal += amp * np.sin(2 * np.pi * f_k * t)
    attack_len = int(0.01 * sr)
    if attack_len < len(t):
        env = np.ones_like(t)
        env[:attack_len] = np.linspace(0.0, 1.0, attack_len)
        signal *= env
    max_val = np.max(np.abs(signal))
    if max_val > 0:
        signal = signal / max_val * 0.75
    return signal

# 5. Veena Synthesizer
def synth_veena(freq, duration, sr=22050):
    t = np.linspace(0, duration, int(sr * duration), endpoint=False)
    signal = np.zeros_like(t)
    num_harmonics = 10
    B = 0.00008
    for k in range(1, num_harmonics + 1):
        f_k = k * freq * np.sqrt(1 + B * k**2)
        if f_k > sr / 2:
            break
        decay_rate = 1.0 + 0.25 * k
        amp = (1.0 / (k ** 1.25)) * np.exp(-decay_rate * t)
        signal += amp * np.sin(2 * np.pi * f_k * t)
    strike_len = int(0.012 * sr)
    if strike_len < len(t):
        env = np.ones_like(t)
        env[:strike_len] = np.linspace(0.0, 1.0, strike_len)
        signal *= env
    max_val = np.max(np.abs(signal))
    if max_val > 0:
        signal = signal / max_val * 0.75
    return signal

# Audio routing helper
def generate_audio(inst, freq, dur, swara=None, octave=None):
    if inst == 'Voice (Teacher)' or inst == 'Voice (Student)':
        voice = 'teacher' if 'Teacher' in inst else 'student'
        swara_sym = swara if swara is not None else 'S'
        swara_sym = swara_sym.replace('⁺', '').replace('⁻', '').strip()
        # Shift the voice target frequency down by 1 octave (divide by 2.0).
        # This translates the keyboard's default Octave 4 to play the voice at its natural Octave 3 (undistorted).
        return extract_manual_vocal_swara(voice, swara_sym, freq / 2.0, dur, octave)
    elif inst == 'Piano':
        return synth_piano(freq, dur)
    elif inst == 'Violin':
        return synth_violin(freq, dur)
    elif inst == 'Human Male Voice':
        return synth_voice(freq, dur, 'male')
    elif inst == 'Human Female Voice':
        return synth_voice(freq, dur, 'female')
    elif inst == 'Sitar':
        return synth_sitar(freq, dur)
    elif inst == 'Veena':
        return synth_veena(freq, dur)
    return synth_piano(freq, dur)

# --- End of DSP Synthesis Engine ---

# Local copies to avoid dependency on previous cell execution state
ref_carnatic_16 = ['S', 'R₁', 'R₂', 'R₃', 'G₁', 'G₂', 'G₃', 'M₁', 'M₂', 'P', 'D₁', 'D₂', 'D₃', 'N₁', 'N₂', 'N₃']
ref_western_16 = ['C', 'C#', 'D', 'D#', 'D', 'D#', 'E', 'F', 'F#', 'G', 'G#', 'A', 'A#', 'A', 'A#', 'B']
ref_western_12 = ['C', 'C#', 'D', 'D#', 'E', 'F', 'F#', 'G', 'G#', 'A', 'A#', 'B']

# Kannada and Devanagari translation maps for scale swaras
swara_kannada_map = {
    'S': 'ಸ', 'R₁': 'ರಿ₁', 'R₂': 'ರಿ₂', 'R₃': 'ರಿ₃',
    'G₁': 'ಗ₁', 'G₂': 'ಗ₂', 'G₃': 'ಗ₃',
    'M₁': 'ಮ₁', 'M₂': 'ಮ₂', 'P': 'ಪ',
    'D₁': 'ದ₁', 'D₂': 'ದ₂', 'D₃': 'ದ₃',
    'N₁': 'ನಿ₁', 'N₂': 'ನಿ₂', 'N₃': 'ನಿ₃'
}
swara_devanagari_map = {
    'S': 'स', 'R₁': 'रि₁', 'R₂': 'रि₂', 'R₃': 'रि₃',
    'G₁': 'ग₁', 'G₂': 'ग₂', 'G₃': 'ग₃',
    'M₁': 'म₁', 'M₂': 'म₂', 'P': 'प',
    'D₁': 'द₁', 'D₂': 'द₂', 'D₃': 'द₃',
    'N₁': 'नि₁', 'N₂': 'नि₂', 'N₃': 'नि₃'
}

def get_swara_frequency(swara, octave):
    try:
        idx = ref_carnatic_16.index(swara)
        western_note = ref_western_16[idx]
        note_index = ref_western_12.index(western_note)
        midi_note = 12 * (octave + 1) + note_index
        return 440.0 * (2.0 ** ((midi_note - 69) / 12.0))
    except Exception as e:
        return 261.63

# Synthesizer and Control widgets (re-added locally to the last cell)
raga_inst_w = widgets.ToggleButtons(
    options=['Piano', 'Violin', 'Human Male Voice', 'Human Female Voice', 'Sitar', 'Veena', 'Voice (Teacher)', 'Voice (Student)'],
    description='Instrument:',
    button_style='',
    tooltips=['Synthetic Piano', 'Synthetic Violin', 'Synthetic Male Voice', 'Synthetic Female Voice', 'Synthetic Sitar', 'Synthetic Veena', 'Vocal Sample Teacher', 'Vocal Sample Student'],
)

raga_octave_w = widgets.ToggleButtons(
    options=[3, 4, 5],
    value=4,
    description='Octave:',
    button_style='',
)

raga_dur_w = widgets.FloatSlider(
    value=0.8,
    min=0.2,
    max=3.0,
    step=0.1,
    description='Note Dur (s):',
    readout_format='.1f',
)

raga_hold_mode_w = widgets.Checkbox(
    value=False,
    description='Hold Note (Loop)',
    layout=widgets.Layout(width='180px')
)

btn_stop_note = widgets.Button(
    description='Stop Note',
    button_style='danger',
    tooltip='Stop the playing loop',
    layout=widgets.Layout(width='80px')
)

raga_num_w = widgets.IntSlider(value=15, min=1, max=72, step=1, description="Raga No:")
raga_num_text_w = widgets.BoundedIntText(value=15, min=1, max=72, step=1, description="Enter No:")
widgets.jslink((raga_num_w, 'value'), (raga_num_text_w, 'value'))

details_out = widgets.HTML()
swara_buttons_container = widgets.VBox(layout=widgets.Layout(margin='5px 0'))
raga_output_w = widgets.Output()

raga_phrase_len_w = widgets.IntSlider(
    value=5,
    min=3,
    max=10,
    step=1,
    description='Phrase Length:',
    layout=widgets.Layout(width='300px')
)

btn_play_random = widgets.Button(
    description='Play Random Phrase',
    button_style='info',
    tooltip='Generate and play a random phrase',
    layout=widgets.Layout(width='170px')
)

melakartha_ragas = [
    "Kanakangi", "Ratnangi", "Ganamurti", "Vanaspati", "Manavati", "Tanarupi",
    "Senavati", "Hanumatodi", "Dhenuka", "Natakapriya", "Kokilapriya", "Rupavati",
    "Gayakapriya", "Vakulamabharanam", "Mayamalavagowla", "Chakravaham", "Suryakantam", "Hatakamabari",
    "Jhankaradhwani", "Natabhairavi", "Keeravani", "Kharaharapriya", "Gourimanohari", "Varunapriya",
    "Mararanjani", "Charukesi", "Sarasangi", "Harikambhoji", "Dheerasankarabharanam", "Naganandini",
    "Yagapriya", "Ragavardhini", "Gangeyabhushani", "Vagadhishwari", "Shulini", "Chalanata",
    "Salagam", "Jalanarayanai", "Jhalavarali", "Navaneetam", "Pavani", "Raghupriya",
    "Gavambhodhi", "Bhavapriya", "Subhapantuvarali", "Shadvidhamargini", "Suvarnangi", "Divyamani",
    "Dhavalambari", "Namanarayani", "Kamavardhini", "Ramapriya", "Gamanashrama",
    "Vishvambhari", "Shyamalangi", "Shanmukhapriya", "Simhendramadhyamam", "Hemavati", "Dharmavati",
    "Nitimati", "Kāntāmaṇi", "Riśabhapriyā", "Latāngi", "Vāchaspati", "Mechakalyāni", "Chitrāmbari", 
    "Sucharitrā", "Jyoti svarupini", "Dhatuvardhini", "Nāsikābhūśaṇi", "Kosalam", "Rasikapriya"
]
ragaByNumber = {}
for i, name in enumerate(melakartha_ragas):
    ragaByNumber[i+1] = {'IAST': name}

def update_raga_display(*args):
    num = raga_num_w.value
    raga_details = ragaByNumber.get(num, {})
    raga_name = raga_details.get('IAST', 'Unknown')
    
    scale_str = get_scale(num)
    swaras = scale_str.split()
    western_notes = [ref_western_16[ref_carnatic_16.index(s)] for s in swaras]
    
    details_html = f"""
    <div style="font-family: Arial, sans-serif; padding: 10px; background-color: #f9f9f9; border-radius: 5px; border-left: 5px solid #2a5298;">
        <h3 style="margin: 0 0 8px 0; color: #2a5298;">Melakarta #{num}: {raga_name}</h3>
        <div style="margin-bottom: 5px;">
            <strong>Carnatic Swaras:</strong> <span style="font-size: 1.15em; font-weight: bold; letter-spacing: 1.5px; color: #d32f2f;">{' '.join(swaras)}</span>
        </div>
        <div>
            <strong>Western Notes:</strong> <span style="font-size: 1.15em; letter-spacing: 1.5px; color: #388e3c;">{' '.join(western_notes)}</span>
        </div>
    </div>
    """
    details_out.value = details_html

def play_single_swara(swara, oct_val):
    with raga_output_w:
        clear_output(wait=True)
        inst = raga_inst_w.value
        dur = raga_dur_w.value
        freq = get_swara_frequency(swara, oct_val)
        
        loop_mode = raga_hold_mode_w.value
        
        if loop_mode:
            matra_dur = 0.45
            sig_single = generate_audio(inst, freq, matra_dur, swara=swara, octave=oct_val)
            sig = np.tile(sig_single, 16)
        else:
            sig = generate_audio(inst, freq, dur, swara=swara, octave=oct_val)
            
        buf = io.BytesIO()
        wavfile.write(buf, 22050, (sig * 32767).astype(np.int16))
        b64_data = base64.b64encode(buf.getvalue()).decode('utf-8')
        
        mode_label = "Loop Mode" if loop_mode else "Single Play"
        print_str = f"Playing note: {swara} ({freq:.2f} Hz) on {inst} (Octave {oct_val}) | {mode_label}"
        
        audio_html = f"""
        <div style="margin-top: 5px;">
            <p style="font-weight: bold; color: #2a5298; margin: 0 0 5px 0;">{print_str}</p>
            <audio autoplay controls style="width: 100%; max-width: 400px;">
                <source src="data:audio/wav;base64,{b64_data}" type="audio/wav">
            </audio>
        </div>
        """
        display(widgets.HTML(audio_html))

def stop_note_action(b):
    with raga_output_w:
        clear_output(wait=True)
        print("Note stopped.")

btn_stop_note.on_click(stop_note_action)

def update_swara_buttons(*args):
    num = raga_num_w.value
    scale_str = get_scale(num)
    scale_notes = scale_str.split()
    
    octaves_list = [3, 4, 5]
    rows = []
    
    for idx, oct_val in enumerate(octaves_list):
        notes_list = [(s, oct_val) for s in scale_notes]
        if idx == len(octaves_list) - 1:
            notes_list.append(('S', oct_val + 1))
            
        row_buttons = []
        for s, o in notes_list:
            is_high = o > oct_val
            suffix = "⁺" if is_high else ""
            btn = widgets.Button(
                description=f"{s}{suffix}",
                layout=widgets.Layout(width='60px', height='40px', margin='2px')
            )
            btn.style.button_color = '#e1f5fe' if is_high else '#f5f5f5'
            btn.style.text_color = '#0288d1' if is_high else 'black'
            btn.style.font_weight = 'bold'
            
            btn.on_click(lambda b, sw=s, oc=o: play_single_swara(sw, oc))
            row_buttons.append(btn)
            
        label = widgets.HTML(
            f"<div style='font-size:0.9em; font-weight:bold; color:#555; width:70px; margin-top:8px;'>Octave {oct_val}:</div>"
        )
        rows.append(widgets.HBox([label] + row_buttons))
        
    swara_buttons_container.children = rows

raga_num_w.observe(update_raga_display, names='value')
raga_num_w.observe(update_swara_buttons, names='value')

# Initialize
update_raga_display()
update_swara_buttons()

def play_raga_scale_action(order_type):
    with raga_output_w:
        clear_output(wait=True)
        num = raga_num_w.value
        raga_details = ragaByNumber.get(num, {})
        raga_name = raga_details.get('IAST', '')
        
        scale_str = get_scale(num)
        swaras = scale_str.split()
        
        inst = raga_inst_w.value
        oct_val = raga_octave_w.value
        dur = raga_dur_w.value
        
        if order_type == 'arohana':
            seq = [(s, oct_val) for s in swaras] + [('S', oct_val + 1)]
            title_label = "Arohana (Ascending)"
        elif order_type == 'avarohana':
            seq = [('S', oct_val + 1)] + [(s, oct_val) for s in reversed(swaras)]
            title_label = "Avarohana (Descending)"
        elif order_type == 'combination':
            seq = [(s, oct_val) for s in swaras] + [('S', oct_val + 1)]
            seq += [('S', oct_val + 1)] + [(s, oct_val) for s in reversed(swaras)]
            title_label = "Scale (Arohana + Avarohana)"
        
        sr = 22050
        signals = [generate_audio(inst, get_swara_frequency(s, o), dur, swara=s, octave=o) for s, o in seq]
        
        combined = concatenate_signals(signals, sr, crossfade_ms=15)
        
        buf = io.BytesIO()
        wavfile.write(buf, sr, (combined * 32767).astype(np.int16))
        b64_data = base64.b64encode(buf.getvalue()).decode('utf-8')
        
        print_str = f"🎼 Playing {title_label} of Raga #{num} ({raga_name}) on {inst}"
        audio_html = f"""
        <div style="margin-top: 5px;">
            <p style="font-weight: bold; color: #2a5298; margin: 0 0 5px 0;">{print_str}</p>
            <audio autoplay controls style="width: 100%; max-width: 400px;">
                <source src="data:audio/wav;base64,{b64_data}" type="audio/wav">
            </audio>
        </div>
        """
        display(widgets.HTML(audio_html))

def play_random_phrase(b):
    with raga_output_w:
        clear_output(wait=True)
        num = raga_num_w.value
        raga_details = ragaByNumber.get(num, {})
        raga_name = raga_details.get('IAST', '')
        
        scale_str = get_scale(num)
        swaras = scale_str.split()
        
        inst = raga_inst_w.value
        octave = raga_octave_w.value
        dur = raga_dur_w.value
        length = raga_phrase_len_w.value
        
        pool = [(s, octave) for s in swaras] + [('S', octave + 1)]
        seq = [random.choice(pool) for _ in range(length)]
        
        sr = 22050
        signals = []
        seq_display = []
        for s, oct_val in seq:
            freq = get_swara_frequency(s, oct_val)
            sig = generate_audio(inst, freq, dur, swara=s, octave=oct_val)
            signals.append(sig)
            
            suffix = "⁺" if oct_val > octave else ""
            seq_display.append(f"{s}{suffix}")
            
        combined = concatenate_signals(signals, sr, crossfade_ms=15)
        
        buf = io.BytesIO()
        wavfile.write(buf, sr, (combined * 32767).astype(np.int16))
        b64_data = base64.b64encode(buf.getvalue()).decode('utf-8')
        
        print_str = f"🎲 Random phrase in Raga #{num} ({raga_name}): {' → '.join(seq_display)}"
        audio_html = f"""
        <div style="margin-top: 5px;">
            <p style="font-weight: bold; color: #2a5298; margin: 0 0 5px 0;">{print_str}</p>
            <audio autoplay controls style="width: 100%; max-width: 400px;">
                <source src="data:audio/wav;base64,{b64_data}" type="audio/wav">
            </audio>
        </div>
        """
        display(widgets.HTML(audio_html))

btn_raga_aro = widgets.Button(description="Arohana", button_style='primary', layout=widgets.Layout(width='120px'))
btn_raga_avar = widgets.Button(description="Avarohana", button_style='primary', layout=widgets.Layout(width='120px'))
btn_raga_comb = widgets.Button(description="Combination", button_style='success', layout=widgets.Layout(width='120px'))

btn_raga_aro.on_click(lambda b: play_raga_scale_action('arohana'))
btn_raga_avar.on_click(lambda b: play_raga_scale_action('avarohana'))
btn_raga_comb.on_click(lambda b: play_raga_scale_action('combination'))
btn_play_random.on_click(play_random_phrase)

raga_ui = widgets.VBox([
    widgets.HTML("<h2>🎶 Melakarta Raga Scale Synthesizer 🎶</h2>"),
    widgets.VBox([
        widgets.HTML("<h4>Instrument & Synthesis Options:</h4>"),
        widgets.HBox([raga_inst_w]),
        widgets.HBox([raga_octave_w, raga_dur_w])
    ], layout=widgets.Layout(margin='5px 0', border='1px solid #ccc', padding='10px', border_radius='8px')),
    widgets.HBox([raga_num_w, raga_num_text_w]),
    details_out,
    widgets.VBox([
        widgets.HTML("<h4>Play Individual Scale Swaras (Click to play):</h4>"),
        widgets.HBox([raga_hold_mode_w, btn_stop_note], layout=widgets.Layout(align_items='center', margin='0 0 10px 0')),
        swara_buttons_container
    ], layout=widgets.Layout(margin='5px 0', border='1px solid #ccc', padding='10px', border_radius='8px')),
    widgets.VBox([
        widgets.HTML("<h4>Scale Playback:</h4>"),
        widgets.HBox([btn_raga_aro, btn_raga_avar, btn_raga_comb])
    ], layout=widgets.Layout(margin='5px 0')),
    widgets.VBox([
        widgets.HTML("<h4>Random Sequence Generator:</h4>"),
        widgets.HBox([raga_phrase_len_w, btn_play_random])
    ], layout=widgets.Layout(margin='5px 0', border='1px solid #ccc', padding='10px', border_radius='8px')),
    widgets.HTML("<h4 style='margin-top: 15px;'>Raga Audio Output:</h4>"),
    raga_output_w
], layout=widgets.Layout(
    max_width='800px',
    padding='15px',
    background_color='#ffffff',
    border_radius='10px',
    box_shadow='0px 4px 15px rgba(0,0,0,0.1)'
))

display(raga_ui)
```

```python
def play_audio_timestamp(*args):
    """Loads and plays a specific window of a long audio file."""
    # print(f'no. of {len(args)} args {args}')
    path = args[0]
    start_min, start_sec, end_min, end_sec = args[1] 
    nam = args[2] if len(args) > 2 else "Unknown"
    # print(f'{start_min}:{start_sec} {end_min}:{end_sec} {nam}')
    t_start = (start_min * 60) + start_sec
    t_end = (end_min * 60) + end_sec
    t_duration = t_end - t_start
    
    # Load segment
    y_seg, sr_seg = librosa.load(path, offset=t_start, duration=t_duration)
    
    print(f"Playing snippet {nam} from {start_min}:{start_sec:02f} to {end_min}:{end_sec:02f}")
    return ipd.Audio(y_seg, rate=sr_seg)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import librosa
import librosa.display
import IPython.display as ipd
from IPython.display import display

def concatenate_swaras(y, sr, swara_list, crossfade_ms=15):
    """
    Concatenates multiple swaras specified as a list of timestamp tuples.
    
    Parameters:
    - y: NumPy array of the loaded audio file.
    - sr: Sampling rate of the audio.
    - swara_list: List of tuples (start_min, start_sec, end_min, end_sec) representing notes.
    - crossfade_ms: Overlap-add duration in milliseconds. Set to 0 for raw concatenation.
    
    Returns:
    - y_concat: The concatenated NumPy audio array.
    - audio_widget: An IPython Audio player for the concatenated audio.
    """
    slices = []
    for t in swara_list:
        t_start = (t[0] * 60) + t[1]
        t_end = (t[2] * 60) + t[3]
        idx_start = int(t_start * sr)
        idx_end = int(t_end * sr)
        slices.append(y[idx_start:idx_end])
        
    if not slices:
        return np.array([]), None
        
    fade_len = int((crossfade_ms * sr) / 1000)
    
    y_concat = slices[0].copy()
    
    for next_slice in slices[1:]:
        if fade_len > 0 and len(y_concat) >= fade_len and len(next_slice) >= fade_len:
            # Linear ramp windows
            fade_out = np.linspace(1.0, 0.0, fade_len)
            fade_in = np.linspace(0.0, 1.0, fade_len)
            
            # Apply fade-out to the end of the current sequence
            y_concat[-fade_len:] = y_concat[-fade_len:] * fade_out
            
            # Copy next slice and apply fade-in to the start
            faded_next = next_slice.copy()
            faded_next[:fade_len] = faded_next[:fade_len] * fade_in
            
            # Overlap-add the faded regions
            overlap = y_concat[-fade_len:] + faded_next[:fade_len]
            
            # Concatenate
            y_concat = np.concatenate([y_concat[:-fade_len], overlap, faded_next[fade_len:]])
        else:
            # Raw concatenation (no crossfade)
            y_concat = np.concatenate([y_concat, next_slice])
            
    return y_concat, ipd.Audio(y_concat, rate=sr)

def compare_original_vs_concatenated(y, sr, original_tuple, swara_list, crossfade_ms=15, title=""):
    """
    Plots the waveform and spectrogram of the original continuous phrase 
    vs. the concatenated sequence of individual notes to explore fidelity.
    
    Parameters:
    - y: NumPy array of the loaded audio file.
    - sr: Sampling rate of the audio.
    - original_tuple: Tuple (start_min, start_sec, end_min, end_sec) representing the continuous phrase.
    - swara_list: List of tuples representing the individual notes to concatenate.
    - crossfade_ms: Crossfade length in milliseconds.
    - title: Title identifier (e.g., 'Teacher Ascent', 'My Descent').
    """
    # Extract original
    t_start_orig = original_tuple[0] * 60 + original_tuple[1]
    t_end_orig = original_tuple[2] * 60 + original_tuple[3]
    idx_start_orig = int(t_start_orig * sr)
    idx_end_orig = int(t_end_orig * sr)
    y_orig = y[idx_start_orig:idx_end_orig]
    
    # Concatenate
    y_concat, _ = concatenate_swaras(y, sr, swara_list, crossfade_ms=crossfade_ms)
    
    print(f"[{title}] Original Duration: {len(y_orig)/sr:.3f} seconds")
    print(f"[{title}] Concatenated Duration: {len(y_concat)/sr:.3f} seconds")
    
    fig, axes = plt.subplots(4, 1, figsize=(14, 12))
    
    # 1. Waveform - Original
    librosa.display.waveshow(y_orig, sr=sr, ax=axes[0], color='dodgerblue')
    axes[0].set_title(f"Original Continuous Phrase: {title}")
    axes[0].set_ylabel("Amplitude")
    axes[0].grid(True, alpha=0.3)
    
    # 2. Waveform - Concatenated
    librosa.display.waveshow(y_concat, sr=sr, ax=axes[1], color='mediumpurple')
    axes[1].set_title(f"Concatenated Notes (crossfade: {crossfade_ms}ms)")
    axes[1].set_ylabel("Amplitude")
    axes[1].grid(True, alpha=0.3)
    
    # Add vertical dashed lines at note boundaries in concatenated audio
    curr_time = 0.0
    for i, t in enumerate(swara_list[:-1]):
        dur = (t[2]*60 + t[3]) - (t[0]*60 + t[1])
        curr_time += dur
        if crossfade_ms > 0:
            curr_time -= (crossfade_ms / 1000.0)
        axes[1].axvline(x=curr_time, color='crimson', linestyle='--', alpha=0.8, label='Boundary' if i==0 else "")
    if len(swara_list) > 1:
        axes[1].legend(loc='upper right')
        
    # 3. Spectrogram - Original
    D_orig = librosa.amplitude_to_db(np.abs(librosa.stft(y_orig)), ref=np.max)
    img_orig = librosa.display.specshow(D_orig, sr=sr, x_axis='time', y_axis='linear', ax=axes[2], cmap='magma')
    axes[2].set_title("Original Spectrogram (showing continuous portamento / pitch glides)")
    axes[2].set_ylim(0, 1500) # vocal frequency range of interest
    fig.colorbar(img_orig, ax=axes[2], format="%+2.0f dB")
    
    # 4. Spectrogram - Concatenated
    D_concat = librosa.amplitude_to_db(np.abs(librosa.stft(y_concat)), ref=np.max)
    img_concat = librosa.display.specshow(D_concat, sr=sr, x_axis='time', y_axis='linear', ax=axes[3], cmap='magma')
    axes[3].set_title("Concatenated Spectrogram (showing discrete frequency jumps at boundaries)")
    axes[3].set_ylim(0, 1500)
    fig.colorbar(img_concat, ax=axes[3], format="%+2.0f dB")
    
    # Add boundary lines to concatenated spectrogram
    curr_time = 0.0
    for i, t in enumerate(swara_list[:-1]):
        dur = (t[2]*60 + t[3]) - (t[0]*60 + t[1])
        curr_time += dur
        if crossfade_ms > 0:
            curr_time -= (crossfade_ms / 1000.0)
        axes[3].axvline(x=curr_time, color='cyan', linestyle='--', alpha=0.8)
        
    plt.tight_layout()
    plt.show()

    print(f"Original Playback ({title}):")
    display(ipd.Audio(y_orig, rate=sr))
    print(f"Concatenated Playback ({title}):")
    display(ipd.Audio(y_concat, rate=sr))
    
    return y_orig, y_concat
```

### Exploration 3: Pitch-Shifting Swaras (ri1 to ri2, ga1 to ga2)

Since the original voice recordings (teacher and student) were sung in the **Mayamalavagowla** scale, they only contain the swara variants `ri1` ($R_1$) and `ga3` ($G_3$) (along with `sa`, `ma1`, `pa`, `da1`, `ni3`, `sa2`). They do not contain other scale variants like `ri2` ($R_2$, Chatushruthi Rishaba), `ga1` ($G_1$, Shuddha Gaandhaara), or `ga2` ($G_2$, Saadhaarana Gaandhaara).

We can synthesize these missing notes by pitch-shifting the recorded segments using the semitone intervals defined in **Section 1**:
- **$R_1$** (Shuddha Rishaba) $\rightarrow$ Western note $C\#$ (semitone 1)
- **$R_2$ / $G_1$** (Chatushruthi Rishaba / Shuddha Gaandhaara) $\rightarrow$ Western note $D$ (semitone 2)
- **$G_2$** (Saadhaarana Gaandhaara) $\rightarrow$ Western note $D\#$ (semitone 3)
- **$G_3$** (Anthara Gaandhaara) $\rightarrow$ Western note $E$ (semitone 4)

Using `librosa.effects.pitch_shift`, we can shift:
- **`ri1` $\rightarrow$ `ri2`** by shifting `+1` semitone.
- **`ri1` $\rightarrow$ `ga1`** by shifting `+1` semitone.
- **`ga3` $\rightarrow$ `ga2`** by shifting `-1` semitone.
- **`ni3` $\rightarrow$ `ni2`** by shifting `-1` semitone.
- **`da1` $\rightarrow$ `da2`** by shifting `+1` semitone.

```python
def get_voice_swara(y, sr, voice, direction, target_swara):
    """
    Retrieves the audio slice for target_swara in the voice (teacher/student) and direction.
    If the swara is already recorded, returns it directly.
    If it is not recorded (e.g. ri2, ga1, ga2, da2, ni2), pitch-shifts the closest available recorded swara
    using the semitone offsets defined in Section 1.
    """
    # Mappings from target swara to (source swara, semitone shift)
    shift_map = {
        'sa': ('sa', 0),
        'ri1': ('ri1', 0),
        'ri2': ('ri1', 1),   # R2 is 1 semitone above R1 (C# to D)
        'ri3': ('ri1', 2),   # R3 is 2 semitones above R1 (C# to D#)
        'ga1': ('ri1', 1),   # G1 is 1 semitone above R1 (C# to D)
        'ga2': ('ga3', -1),  # G2 is 1 semitone below G3 (E to D#)
        'ga3': ('ga3', 0),
        'ma1': ('ma1', 0),
        'ma2': ('ma1', 1),   # M2 is 1 semitone above M1 (F to F#)
        'pa': ('pa', 0),
        'da1': ('da1', 0),
        'da2': ('da1', 1),   # D2 is 1 semitone above D1 (G# to A)
        'da3': ('da1', 2),   # D3 is 2 semitones above D1 (G# to A#)
        'ni1': ('ni3', -2),  # N1 is 2 semitones below N3 (B to A)
        'ni2': ('ni3', -1),  # N2 is 1 semitone below N3 (B to A#)
        'ni3': ('ni3', 0),
        'sa2': ('sa2', 0)
    }
    
    if target_swara not in shift_map:
        raise ValueError(f"Unknown target swara: {target_swara}")
        
    src_swara, semitones = shift_map[target_swara]
    
    # Retrieve segment timestamp from consolidated swara_metadata
    try:
        t = get_swara_timestamp(voice, direction, src_swara, type_val="note")
    except ValueError:
        # Fallback to alternative direction if not found
        alt_dir = "ascending" if direction == "descending" else "descending"
        t = get_swara_timestamp(voice, alt_dir, src_swara, type_val="note")
        
    t_start = (t[0] * 60) + t[1]
    t_end = (t[2] * 60) + t[3]
    idx_start = int(t_start * sr)
    idx_end = int(t_end * sr)
    y_seg = y[idx_start:idx_end].copy()
    
    if semitones != 0:
        import librosa
        y_seg = librosa.effects.pitch_shift(y_seg, sr=sr, n_steps=semitones)
        
    return y_seg

def concatenate_voice_swaras(y, sr, voice, direction, swara_list, crossfade_ms=15):
    """
    Concatenates voice swaras, dynamically generating pitch-shifted segments when notes
    are not directly available in the original recordings.
    """
    import numpy as np
    import IPython.display as ipd
    
    slices = []
    for swara in swara_list:
        y_seg = get_voice_swara(y, sr, voice, direction, swara)
        slices.append(y_seg)
        
    if not slices:
        return np.array([]), None
        
    fade_len = int((crossfade_ms * sr) / 1000)
    y_concat = slices[0].copy()
    
    for next_slice in slices[1:]:
        if fade_len > 0 and len(y_concat) >= fade_len and len(next_slice) >= fade_len:
            fade_out = np.linspace(1.0, 0.0, fade_len)
            fade_in = np.linspace(0.0, 1.0, fade_len)
            
            y_concat[-fade_len:] = y_concat[-fade_len:] * fade_out
            faded_next = next_slice.copy()
            faded_next[:fade_len] = faded_next[:fade_len] * fade_in
            
            overlap = y_concat[-fade_len:] + faded_next[:fade_len]
            y_concat = np.concatenate([y_concat[:-fade_len], overlap, faded_next[fade_len:]])
        else:
            y_concat = np.concatenate([y_concat, next_slice])
            
    return y_concat, ipd.Audio(y_concat, rate=sr)
```

### Exploration 4: Natural Octave Swaras vs. Artificial Pitch-Shifting (Resolving 'Chipmunk' Effects)

When we pitch-shift a singer's voice artificially (e.g. shifting middle-octave notes up or down by an octave), it shifts both the fundamental frequency (pitch) and the spectral envelope (formants). Shifting formants upwards makes the vocal tract sound physically smaller, resulting in the **artificial chipmunk effect**. Shifting downwards makes it sound unnaturally deep/monster-like.

To resolve this, we use **naturally sung notes** from different octaves (low and high) recorded under the `octave_low` and `octave_high` dictionaries. In natural singing, the singer changes pitch by adjusting vocal cord tension, while keeping their vocal tract shape relatively constant. This keeps the formants in their natural frequency bands, preserving the original vocal quality.

Below, we define the `octave_low` and `octave_high` dictionaries containing the manual segment timestamps for low-octave and high-octave notes.

```python
# 2. Extract and compare naturally sung swaras vs artificially pitch-shifted swaras

# Timestamps of corresponding middle octave notes from Experiments-01 (Teacher/Amma)
middle_notes = {
    "ri1": (3, 45.76, 3, 47.24),
    "ga3": (3, 47.25, 3, 48.59),
    "ma1": (3, 48.79, 3, 50.59),
    "pa": (3, 56.2, 3, 57.6),
    "da1": (3, 57.72, 3, 59.2),
    "ni3": (3, 59.3, 4, 0.5)
}

def extract_swara_slice(y, sr, t_tuple):
    t_start = (t_tuple[0] * 60) + t_tuple[1]
    t_end = (t_tuple[2] * 60) + t_tuple[3]
    return y[int(t_start * sr):int(t_end * sr)]
```

### Exploration 5: Synthesizing Unrecorded Octave Swaras (V3 Refined)

Iterative listening refinements addressed four specific acoustic problems:

| Issue | Root Cause | V3 Fix |
| :--- | :--- | :--- |
| **Option 1 squeaky** | Cepstral filter over-amplified $>3.5\text{ kHz}$ spectral residuals | 6 dB/oct shelf de-emphasis above $3.5\text{ kHz}$ |
| **Option 1 Ni3 too quiet** | Normalised to middle-octave RMS, not high-octave reference | Re-normalise output RMS to match adjacent high-octave segment |
| **Option 2 Da1 chipmunkish** | Full $160\text{ ms}$ attack shifted $+12$ semitones including voiced vowel glide | Split into: unvoiced burst ($60\text{ ms}$, naive shift) + voiced glide ($100\text{ ms}$, formant-preserved shift) |
| **Option 2 Ni3 'Sa' residual** | Donor `ri1` was sliced at $160\text{ ms}$ — still inside its own consonant transition | Extended `donor_skip_ms` to $220\text{ ms}$ to land in pure $/i/$ steady-state sustain; looped sustain to match target duration |

```python
import numpy as np
import scipy.signal
import librosa
import IPython.display as ipd
from IPython.display import display

def pitch_shift_formant_preserved_v3(y_sig, sr, n_steps=12, n_fft=2048, hop_length=512,
                                      lifter_cutoff=18, ref_segment=None):
    """
    Option 1 V3:
    - Cepstral formant compensation with high-frequency de-emphasis shelf above 3.5 kHz (removes squeakiness).
    - Ghost-fundamental suppression below 350 Hz (removes double-voice).
    - RMS-matched output amplitude to adjacent high-octave reference segment.
    """
    y_shifted = librosa.effects.pitch_shift(y_sig, sr=sr, n_steps=n_steps, n_fft=n_fft, hop_length=hop_length)

    def get_envelope(sig):
        D = librosa.stft(sig, n_fft=n_fft, hop_length=hop_length)
        lm = np.log(np.abs(D) + 1e-8)
        c = np.fft.ifft(lm, axis=0)
        lif = np.zeros_like(c); lif[:lifter_cutoff] = c[:lifter_cutoff]; lif[-lifter_cutoff:] = c[-lifter_cutoff:]
        return np.exp(np.real(np.fft.fft(lif, axis=0))), D

    env_orig, _ = get_envelope(y_sig)
    env_shift, D_shift = get_envelope(y_shifted)

    mt = min(env_orig.shape[1], env_shift.shape[1], D_shift.shape[1])
    env_orig = env_orig[:, :mt]; env_shift = env_shift[:, :mt]; D_shift = D_shift[:, :mt]

    filt = np.clip(env_orig / (env_shift + 1e-6), 0.2, 5.0)
    freqs = librosa.fft_frequencies(sr=sr, n_fft=n_fft)
    
    # Low-frequency subharmonic neutralisation (anti double-voice)
    lo_bin = np.searchsorted(freqs, 350.0)
    for i in range(lo_bin):
        w = float(i) / lo_bin
        filt[i, :] = 1.0 * (1 - w) + filt[i, :] * w

    # High-frequency shelf de-emphasis above 3.5 kHz (anti squeaky)
    hi_bin = np.searchsorted(freqs, 3500.0)
    n_bins = len(freqs)
    for i in range(hi_bin, n_bins):
        shelf = 1.0 - 0.4 * min(1.0, (i - hi_bin) / max(1, n_bins - hi_bin))
        filt[i, :] *= shelf

    D_comp = D_shift * filt
    _, y_out = scipy.signal.istft(D_comp, fs=sr, nperseg=n_fft, noverlap=n_fft - hop_length)

    sos = scipy.signal.butter(4, 250, 'hp', fs=sr, output='sos')
    y_out = scipy.signal.sosfilt(sos, y_out)

    if len(y_out) > len(y_shifted): y_out = y_out[:len(y_shifted)]
    elif len(y_out) < len(y_shifted): y_out = np.pad(y_out, (0, len(y_shifted) - len(y_out)))

    # RMS normalise to reference high-octave level
    ref = ref_segment if ref_segment is not None else y_sig
    ref_rms = np.sqrt(np.mean(ref**2)) + 1e-9
    out_rms = np.sqrt(np.mean(y_out**2)) + 1e-9
    return y_out * (ref_rms / out_rms)


def synthesize_swara_consonant_transplant_v3(
    y_middle_src, y_high_neighbor, sr,
    n_steps_shift=1, burst_ms=60, voiced_glide_ms=100,
    donor_skip_ms=200, crossfade_ms=40, n_fft=2048, hop_length=512
):
    """
    Option 2 V3:
    - Splits onset into unvoiced burst (naive +12 st shift) + voiced glide (formant-preserved shift).
    - Skips deep into donor note (donor_skip_ms) to land in pure steady-state vowel sustain.
    - Loops sustain to match source note duration.
    - RMS normalises to donor amplitude.
    """
    y_high_shifted = librosa.effects.pitch_shift(y_high_neighbor, sr=sr, n_steps=n_steps_shift) if n_steps_shift != 0 else y_high_neighbor.copy()

    burst_s = min(int(burst_ms * sr / 1000), len(y_middle_src) // 4)
    glide_s = min(int(voiced_glide_ms * sr / 1000), len(y_middle_src) - burst_s)
    skip_s  = int(donor_skip_ms * sr / 1000)
    fade_s  = int(crossfade_ms * sr / 1000)

    # 2a. Unvoiced burst: safe naive shift
    onset_burst = librosa.effects.pitch_shift(y_middle_src[:burst_s].copy(), sr=sr, n_steps=12)

    # 2b. Voiced glide: formant-preserved shift
    def fp_shift_short(sig):
        if len(sig) < n_fft // 2:
            return librosa.effects.pitch_shift(sig, sr=sr, n_steps=12)
        y_sh = librosa.effects.pitch_shift(sig, sr=sr, n_steps=12, n_fft=n_fft, hop_length=hop_length)
        def env(s):
            D = librosa.stft(s, n_fft=n_fft, hop_length=hop_length)
            lm = np.log(np.abs(D) + 1e-8)
            c = np.fft.ifft(lm, axis=0)
            lif = np.zeros_like(c); lif[:18] = c[:18]; lif[-18:] = c[-18:]
            return np.exp(np.real(np.fft.fft(lif, axis=0))), D
        eo, _ = env(sig); es, Ds = env(y_sh)
        mt = min(eo.shape[1], es.shape[1], Ds.shape[1])
        filt = np.clip(eo[:,:mt] / (es[:,:mt] + 1e-6), 0.2, 4.0)
        _, yo = scipy.signal.istft(Ds[:,:mt] * filt, fs=sr, nperseg=n_fft, noverlap=n_fft-hop_length)
        return yo[:len(y_sh)] if len(yo) >= len(y_sh) else np.pad(yo, (0, len(y_sh)-len(yo)))

    onset_glide = fp_shift_short(y_middle_src[burst_s:burst_s + glide_s].copy())

    # Crossfade burst -> glide
    mf = min(fade_s // 2, burst_s // 4, glide_s // 4)
    ov = onset_burst[-mf:] * np.linspace(1,0,mf) + onset_glide[:mf] * np.linspace(0,1,mf)
    onset_attack = np.concatenate([onset_burst[:-mf], ov, onset_glide[mf:]])

    # 3. Donor sustain — skip deep, then loop to fill target duration
    sustain_raw = y_high_shifted[skip_s:] if len(y_high_shifted) > skip_s + fade_s else y_high_shifted[len(y_high_shifted)//3:]
    target_sustain_len = len(y_middle_src) - len(onset_attack) + fade_s
    if target_sustain_len > len(sustain_raw):
        reps = int(np.ceil(target_sustain_len / len(sustain_raw))) + 1
        sustain_raw = np.tile(sustain_raw, reps)
    sustain_body = sustain_raw[:target_sustain_len]

    # 4. Crossfade splice attack -> sustain
    fo = np.linspace(1.0, 0.0, fade_s); fi = np.linspace(0.0, 1.0, fade_s)
    ov2 = onset_attack[-fade_s:] * fo + sustain_body[:fade_s] * fi
    y_comp = np.concatenate([onset_attack[:-fade_s], ov2, sustain_body[fade_s:]])

    # RMS normalise to donor amplitude
    ref_rms = np.sqrt(np.mean(y_high_neighbor**2)) + 1e-9
    out_rms = np.sqrt(np.mean(y_comp**2)) + 1e-9
    return y_comp * (ref_rms / out_rms)
```
